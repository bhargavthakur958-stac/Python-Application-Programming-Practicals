```python
#problem 1 
class Person:
    def __init__(self, name, id):
        self.name = name
        self.id = id

    def Display(self):
        print("Name:", self.name)
        print("ID:", self.id)


class Emp(Person):
    def __init__(self, name, id):
        super().__init__(name, id)

    def Print(self):
        print("Employee class called")


Emp_details = Emp("Bhargav Thakur", 103)


Emp_details.Display()


Emp_details.Print()
```

```python
# problem 2
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

if sorted(str1.lower()) == sorted(str2.lower()):
    print("Strings are Anagrams")
else:
    print("Strings are not Anagrams")
```

```python
# problem 3
num = int(input("Enter a number: "))

temp = num
power = len(str(num))
sum = 0

while temp > 0:
    digit = temp % 10
    sum = sum + digit ** power
    temp = temp // 10

if sum == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
# problem 4
class Vehicle:

    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.color = "White"

    def seating_capacity(self, capacity):
        self.capacity = capacity

    def display(self):
        print("Maximum Speed:", self.max_speed, "kmph")
        print("Mileage:", self.mileage, "kmpl")
        print("Color:", self.color)
        print("Seating Capacity:", self.capacity)
        print("------------------------")



car1 = Vehicle(200, 20)
car1.seating_capacity(5)


car2 = Vehicle(180, 25)
car2.seating_capacity(4)


print("First Vehicle Details:")
car1.display()

print("Second Vehicle Details:")
car2.display()
```
