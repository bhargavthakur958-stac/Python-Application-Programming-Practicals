```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python

from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

df['species'] = pd.Categorical.from_codes(
    iris.target, iris.target_names
)

print(df.head())

print(df.info())

print(df.describe())
```

```python
# Scatter Plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='sepal length (cm)',
    y='petal length (cm)',
    hue='species',
    s=80
)

plt.title('Scatter Plot of Iris Dataset')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.show()
```

```python
# Density Plot
plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=df,
    x='petal length (cm)',
    hue='species',
    fill=False,
    linewidth=2
)

plt.title('Density Plot of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Density')
plt.show()
```

```python
# Violin Plot
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x='species',
    y='petal length (cm)'
)

plt.title('Violin Plot of Petal Length')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.show()
```

```python
# Calculate correlation
correlation = df.select_dtypes(include='number').corr()

# Create Heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Iris Dataset')
plt.show()
```

```python
# Joint Plot
sns.jointplot(
    data=df,
    x='sepal length (cm)',
    y='petal length (cm)',
    kind='scatter',
    height=7
)

plt.show()
```

```python
# Pair Plot
sns.pairplot(
    df,
    hue='species',
    diag_kind='hist'
)

plt.suptitle('Pair Plot of Iris Dataset', y=1.02)
plt.show()
```
