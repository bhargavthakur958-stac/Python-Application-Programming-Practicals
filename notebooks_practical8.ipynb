```python
# problem 4 
import pandas as pd
import numpy as np

data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Amit', 'Riya', 'Rahul', 'Priya', 'Karan',
             'Neha', 'Vikas', 'Simran', 'Rohit', 'Anjali'],
    'Age': [25, 32, 28, 35, 42, 29, 38, 24, 45, 31],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Purchase_Amount': [5000, 7000, 4500, 8000, 6000,
                        5500, 9000, 4000, 7500, 6500]
}

df = pd.DataFrame(data)

print("First 5 Rows:")
print(df.head())

print("\nSummary Statistics:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].median()
)

print("\nData After Handling Missing Values:")
print(df)
```
